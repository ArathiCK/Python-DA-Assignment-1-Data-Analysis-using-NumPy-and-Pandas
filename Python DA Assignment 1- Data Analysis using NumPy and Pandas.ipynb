## Step 3: Manipulating the DataFrame

```python
# Modify Amount for TransactionID 102 to 165
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'] = 165

m=pd.DataFrame(transactions)
m
```

```python
# Add a new column Discount by calculating 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10

c=pd.DataFrame(transactions)
c
```

```python
# Remove the row with TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109
]

newr=pd.DataFrame(transactions)
newr
```

```python
# Delete the Discount column
transactions = transactions.drop(columns=['Discount'])

d=pd.DataFrame(transactions)
d
```
